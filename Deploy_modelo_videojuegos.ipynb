```python
# Cargamos los datos futuros
data = pd.read_csv('videojuegos-datosFuturos.csv')
data.head()
```

```python
# Creamos variables dummy para convertir la categoría a números
data_preparada = data.copy()
data_preparada = pd.get_dummies(data_preparada, columns=['videojuego', 'Plataforma'], drop_first=False, dtype=int)
data_preparada = pd.get_dummies(data_preparada, columns=['Sexo', 'Consumidor_habitual'], drop_first=False, dtype=int)
data_preparada.head()
```

```python
# Se adicionan la scolumnas faltantes
data_preparada = data_preparada.reindex(columns = variables, fill_value=0)
data_preparada.head()
```

**4. Predicción**
